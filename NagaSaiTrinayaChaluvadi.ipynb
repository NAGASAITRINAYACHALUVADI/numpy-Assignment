**Question:** Why are NumPy arrays generally preferred over plain Python lists for numerical computation in AI/ML workloads? Mention at least two concrete reasons.

**Data Preprocessing Pipeline**

You are given a raw 1D or 2D NumPy array containing NaN values.

Describe, step by step, the NumPy operations you would apply to:
- (a) detect missing values,
- (b) handle them appropriately and justify your strategy, and
- (c) normalize or standardize the clean data before it is fed to a model.

Implement the full pipeline in code on a sample array of your own choosing, including at least one NaN.

````python
data = np.array([10, 20, np.nan, 40, 50], dtype=float)

# Step 1: Detect missing values
print("Missing values:", np.isnan(data).sum())

# Step 2: Handle missing value using mean imputation
data[np.isnan(data)] = np.nanmean(data)

# Step 3: Standardize
standardized = (data - np.mean(data)) / np.std(data)

print("Clean data:", data)
print("Standardized data:", standardized)
````

**Debug the Model Code**

The following code contains more than one bug:

```python
X = np.array([[1, 2],
              [2, 3],
              [3, 4]])  # shape (3, 2)
w = np.array([0.5, 1.0, 0.2])  # intended weight vector
b = 2.0
y_pred = X @ w + b
y_true = np.array([10, 20, 30])
mse = np.mean(y_true - y_pred ** 2)
```

Identify every bug. Explain, in terms of shapes and operator precedence, why each is a problem, and rewrite the corrected code.

**Bug 1 — Shape mismatch:

X.shape = (3, 2)
w.shape = (3,)

X has 2 columns, so the weight vector must contain 2 weights. But w contains 3 values. Therefore, X @ w cannot be performed.

Correct:

w = np.array([0.5, 1.0])

Bug 2 — Operator precedence in MSE:

The original code is:

mse = np.mean(y_true - y_pred ** 2)

Because ** has higher precedence than -, Python evaluates it as:

y_true - (y_pred ** 2)

But MSE requires the difference between actual and predicted values to be squared**

````python
X = np.array([[1, 2],
              [2, 3],
              [3, 4]])

w = np.array([0.5, 1.0])
b = 2.0

y_pred = X @ w + b

y_true = np.array([10, 20, 30])

mse = np.mean((y_true - y_pred) ** 2)

print("Predicted values:", y_pred)
print("MSE:", mse)
````

**Question:** 

Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns.

````python

X = np.array([[10, 100],
              [20, 200],
              [30, 300],
              [40, 400]])

min_values = np.min(X, axis=0)
max_values = np.max(X, axis=0)

X_normalized = (X - min_values) / (max_values - min_values)

print(X_normalized)
````
